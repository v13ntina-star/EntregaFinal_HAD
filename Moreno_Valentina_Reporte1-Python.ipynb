*Importación de las bibliotecas*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

*Lectura y exploración de datos*

```python
df = pd.read_csv ("/content/Poblacion_nacional(in).csv")

#Exploración de los datos
print(df.head()) #Primeras cinco observaciones
print(df.columns.tolist()) #Nombre de las columnas, tolist mejora el formato
print(df.shape) #Cantidad de filas y columnas
print(df.dtypes) #Tipos de datos
print(df.isnull().sum()) #Valores faltantes
```

*Preparación de la serie*

```python
df["Fecha"] = pd.to_datetime(df["Fecha"]) #Identificar la columna que contiene la fecha
print(df["Fecha"])
df.sort_values(by = "Fecha", inplace = True)#Ordenar cronológicamente las observaciones
print(df["Fecha"])
print(df["Valor"].head()) #La columna que contiene los valores de la serie se llama "Valor"
```

*Elaboración y presentación del gráfico*

```python
plt.figure(figsize=(10, 6)) #Tamaño del gráfico
plt.plot(df['Fecha'], df['Valor'], label='Población Nacional') #Leyenda explicativa
plt.title('Evolución y Expectativas de la Población Nacional Chilena') #Definición del titulo descriptivo
plt.xlabel('Año') #Nombre eje horizontal
plt.ylabel('Población Nacional') #Nombre eje vertical
plt.legend()
plt.figtext(0.99, 0.02, 'Fuente: CSV de Poblacion nacional', ha='right') #Fuente de información
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

*Interpretación final*

El gráfico representa la demografía chilena, con datos históricos y las expectativas sobre el nivel poblacional de la nación.
Los datos se distribuyen en el tiempo en forma de campana, con una gran alza desde el año 2000, y una pendiente que hasta el 2020 se mantiene aproximadamente constante, disminuyendo paulatinamente hasta el peak esperado para el año 2030 - 2040, donde se alcanza un valle con una población que rodea los 20 millones, luego de esto, las proyecciones indican una fuerte caida, lo que por un lado, demuestra que la población chilena disminuye, y además, envejece.
Los datos indican que la población se acerca aceleradamente a un declive importante, y de acuerdo a los datos proporcionados por el INE, no existe información suficiente para respaldar que el patron alza-disminución pueda repetirse, aspirando a una baja constante en el futuro.
